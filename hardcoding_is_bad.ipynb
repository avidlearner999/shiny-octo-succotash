```python
# BAD: The discount rate 0.10 is hard-coded all over the program
def apply_discount(price):
    return price - (price * 0.10)
def bulk_discount(prices):
    return [p - (p * 0.10) for p in prices]
def final_price(price, tax_rate):
    discounted = price - (price * 0.10)
    return discounted + (discounted * tax_rate)
print(apply_discount(100))
print(bulk_discount([50, 75, 120]))
print(final_price(200, 0.07))
```

```python
# GOOD: One source of truth
DISCOUNT_RATE = 0.10
def apply_discount(price):
    return price - (price * DISCOUNT_RATE)
def bulk_discount(prices):
    return [p - (p * DISCOUNT_RATE) for p in prices]
def final_price(price, tax_rate):
    discounted = price - (price * DISCOUNT_RATE)
    return discounted + (discounted * tax_rate)
print(apply_discount(100))
```
